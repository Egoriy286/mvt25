   Задача одномерная и выглядит следующим образом:
#### $  0= -k \frac{d^2u}{dx^2}  $, 
#### $ u(0,t) = 0, \hspace{10px}  u(l,t) = 0, \hspace{10px}   $, 
#### где $k$ - коэффициент теплопроводности, $l$ - длина стержня, $t$ - время, $x$ - координата на стержне.
## Дискретизация конечно-разностный метод
#### 

```python
import numpy as np

def construct_matrix(N):
    r = -1
    A = np.zeros((N-1, N-1))

    for i in range(N-1):
        A[i, i] = - 2 * r
        if i > 0:
            A[i, i-1] = r
        if i < N-2:
            A[i, i+1] = r
    return A

# Параметры задачи
l = 1.0        # Длина стержня
#k = -0.1        # Коэффициент теплопроводности
N = 512         # Число узлов по пространству
h = l / N

# Конструируем матрицу
A = construct_matrix(N+1)
print(A.shape[0])
# Печать результатов
print("Матрица A:")
print(A)

# Вектор начальных условий
x = np.linspace(0, l, N)

b = np.zeros(N) 

def save(filename, A, b):
    """
    Сохранение матрицы и вектора в файл.
    :param filename: Имя файла.
    :param A: Матрица (numpy.ndarray).
    :param b: Вектор (numpy.ndarray).
    :param x: Вектор (numpy.ndarray).
    """
    with open(filename, 'w') as f:
        # Запись размерности матрицы
        f.write(f"{A.shape[0]}\n")
        # Запись матрицы
        for row in A:
            f.write(" ".join(map(str, row)) + "\n")
        # Запись вектора
        f.write(" ".join(map(str, b)) + "\n")
        f.write(" ".join(map(str, x)) + "\n")

b[0] = 1
b[-1] = 0
print(b)
save("matrix.txt", A, b)
x = np.linalg.solve(A, b)
for i in range(len(x)):
    print(f" {x[i]:.10f}")
```
